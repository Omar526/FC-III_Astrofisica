Como primera clase, repasaremos conceptos básicos de Python tomando como caso de estudio la cinemática de lanzamiento de proyectiles en 2D. En este problema, un objeto es lanzado formando ángulo con respecto al suelo y sigue una trayectoria en forma de parábola debido a la influencia de la gravedad. Vamos a explorar los conceptos esenciales de este fenómeno y utilizar la física computacional para visualizar la trayectoria.

![lanzamiento de proyectil](proyectil.png)

### **Conceptos Básicos**

- **Velocidad Inicial ($v_0$):** La velocidad con la que el proyectil es lanzado.
- **Ángulo de Lanzamiento ($\theta$):** El ángulo entre la dirección de lanzamiento y el suelo.
- **Aceleración Gravitatoria ($g$):** La aceleración debida a la gravedad, que afecta el movimiento del proyectil.

En este ejercicio, utilizaremos ecuaciones del movimiento parabólico para calcular y visualizar la trayectoria del proyectil. La interacción entre estos parámetros nos permitirá comprender cómo afectan la forma y alcance de la trayectoria.

¡Comencemos explorando este fenómeno!

```python
# Primero importamos las librerías necesarias para nuestra tarea

import numpy as np  # modulo para trabajar con arrays y operaciones matemáticas
import matplotlib.pyplot as plt  # librería para generar gráficas
from matplotlib.animation import FuncAnimation  # función de matplotlib.animation que permite crear animaciones

# recuerda que todo el texto que sigue de un símbolo gato (#) es un comentario
# el interprete de python simplemente omite estas líneas en la ejecución del código
# Los comentarios sirven para poner anotaciones y/o breves explicaciones útiles que describen a tu código

# ES UNA MUY BUENA PRÁCTICA INCLUIR COMENTARIOS, PUEDE QUE EN UN FUTURO
# NECESITES USAR UN CODIGO ANTIGUO Y SABER COMO FUNCIONA
# O PUEDE QUE NECESITES COMPARTIRLO CON TU EQUIPO DE TRABAJO

# LOS COMENTARIOS LES AYUDARÁN A ENTENDER EFICAZMENTE LA ESTRUCURA Y FUNCIONALIDADES DE TU CÓDIGO 
```

Definimos acontinuación los parámetros de velocidad inicial $v_0$, ángulo de lanzamiento $\theta$ y aceleración de gravedad $g$:

```python
# Parámetros iniciales
v0 = 20  # velocidad inicial en m/s
theta = np.radians(80)  # ángulo de lanzamiento en radianes
g = 9.81  # aceleración debido a la gravedad en m/s^2
```

Calculamos el tiempo de vuelo, y con ello la trayectoria del objeto:

```python
# Cálculos de la trayectoria
t_total = 2 * v0 * np.sin(theta) / g  # tiempo total de vuelo
t = np.linspace(0, t_total, num=500)  # valores de tiempo
x = v0 * np.cos(theta) * t  # posición x
y = v0 * np.sin(theta) * t - 0.5 * g * t**2  # posición y
```

Ecuaciones de movimiento:

$$ x(t) = v_0 \cos (\theta) t $$

$$ y(t) = v_0 \sin(\theta) t - \frac{1}{2} g t^2 $$

```python
%%capture 
# El comando "%%capture" evita mostrar el gráfico que se genera en esta celda

# Configuración de la animación
fig, ax = plt.subplots()
ax.set_xlim(0, max(x))
ax.set_ylim(0, max(y) + 5)

line, = ax.plot([], [], lw=2)

def update(frame):
    line.set_data(x[:frame], y[:frame])
    return line,

anim = FuncAnimation(fig, update, frames=len(t), interval=5, blit=True)
```

### **¿Cómo cambia la trayectoria si modificamos el ángulo de lanzamiento?**

```python
# Parámetros iniciales

angles_degree = [0, 30, 45, 60, 90]
angles = np.radians([0, 30, 45, 60, 90])  # ángulos de lanzamiento en radianes
v0 = 20  # velocidad inicial en m/s


fig, ax = plt.subplots()

for angle, angle_degree in zip(angles, angles_degree):
    # Cálculos de la trayectoria
    t_total = 2 * v0 * np.sin(angle) / g  # tiempo total de vuelo
    t = np.linspace(0, t_total, num=1000)  # valores de tiempo
    x = v0 * np.cos(angle) * t  # posición x
    y = v0 * np.sin(angle) * t - 0.5 * g * t**2  # posición y


    #ax.plot(x,y, label=f"{np.round(angle, decimals=2)} rad")
    ax.plot(x,y, label=r"$\theta = $" + f"{angle_degree:.0f}°")

# Ajustar ejes para tener proporción igual
ax.set_aspect('equal', adjustable='box')

ax.set_xlabel(r"$x$ [m]")
ax.set_ylabel(r"$y$ [m]")

ax.legend()

ax.grid()

plt.show()
```

Ahora modificaremos la velocidad inicial del lanzamiento:

```python
# Parámetros iniciales
angle = np.radians(45)  # ángulo de lanzamiento en radianes
#v0s = [0, 10, 20, 30]  # velocidades iniciales en m/s

v0s = range(20)  # velocidades iniciales en m/s

fig, ax = plt.subplots()

for v0 in v0s:
    # Cálculos de la trayectoria
    t_total = 2 * v0 * np.sin(angle) / g  # tiempo total de vuelo
    t = np.linspace(0, t_total, num=1000)  # valores de tiempo
    x = v0 * np.cos(angle) * t  # posición x
    y = v0 * np.sin(angle) * t - 0.5 * g * t**2  # posición y

    ax.plot(x,y)

# Ajustar ejes para tener proporción igual
ax.set_aspect('equal', adjustable='box')

ax.set_xlabel(r"$x$ [m]")
ax.set_ylabel(r"$y$ [m]")

ax.grid()

plt.show()
```

$$ x(t) = \frac{v_{0x}}{b}(1 - \exp(-bt)) $$
