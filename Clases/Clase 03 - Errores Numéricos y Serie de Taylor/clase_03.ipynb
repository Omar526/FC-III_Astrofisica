### **Expansión en Serie de Taylor**

Una expansión en serie de Taylor es una forma de aproximar una función matemática *"compleja"* utilizando una suma de términos polinómicos más simples. Esta herramienta es fundamental tanto en análisis matemático como en computación, ya que permite transformar funciones difíciles de evaluar en expresiones manejables.

La idea central es descomponer la función en una suma infinita de términos, donde cada uno se construye a partir de las derivadas de la función evaluadas en un punto específico \( x_0 \), llamado **punto de expansión**:

$$
f(x) = \sum_{n=0}^{\infty} \frac{f^{(n)}(x_0)}{n!} (x-x_0)^n 
= a_0 + a_1 (x-x_0) + a_2 (x-x_0)^2 + a_3 (x-x_0)^3 + \cdots
$$

El punto \( x_0 \) representa el valor alrededor del cual se construye la aproximación, y es donde la serie reproduce exactamente el comportamiento local de la función. A medida que se agregan más términos, la aproximación mejora en un entorno cercano a este punto.

Un caso particular muy utilizado es cuando \( x_0 = 0 \), lo que da lugar a la **serie de Maclaurin**:

$$
f(x) = \sum_{n=0}^{\infty} \frac{f^{(n)}(0)}{n!} x^n 
= a_0 + a_1 x + a_2 x^2 + a_3 x^3 + \cdots
$$

Calcule el error cuadrático medio para cada $N$ (RMSE, *Root Mean Square Error*), definido como:

$$ \text{RMSE} = \sqrt{ \sum_{i=1}^N \frac{(\hat{y}_i - y_i)^2}{N} } $$

donde:

- $N$ es el número total de observaciones o puntos de datos.
- $y_i$ corresponde al valor real o exacto en el punto $i$.
- $\hat{y}_i$ corresponde al valor aproximado o predicho en el punto $i$.

El RMSE mide la magnitud promedio del error entre los valores aproximados y los valores reales. En el contexto de métodos numéricos, permite cuantificar qué tan precisa es una aproximación computacional respecto a la solución exacta, penalizando más fuertemente los errores grandes debido al uso del cuadrado.

**Conclusión:** ¿Qué aprendimos?

1. Existen dos tipos de errores numéricos comunes en los cálculos computacionales:

    * Error de redondeo: Surge debido a la cantidad limitada de bits disponibles para representar los números en la computadora.

    * Error de aproximación: Ocurre cuando se emplean métodos o algoritmos matemáticos que solo ofrecen una solución aproximada al problema.

2. El orden en que se suman los términos sí influye en el resultado final de una suma en el contexto computacional.

3. En una serie numérica, aumentar el número de términos $N$ no siempre reduce el error total. Aunque un $N$ mayor disminuye el error de aproximación, también puede incrementar el error de redondeo acumulativo, afectando la precisión del resultado.
