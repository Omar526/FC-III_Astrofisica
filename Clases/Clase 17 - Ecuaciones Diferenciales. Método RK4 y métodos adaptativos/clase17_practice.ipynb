### **Método de Runge-Kutta de orden 4**

Considere una ecuación diferencial ordinaria de primer orden en su forma estándar:

$$ \frac{dy}{dt} = f(t, y) $$

El método de Runge-Kutta de orden 4 (RK4) estima $y_{n+1}$ a partir de $y_n$ y $t_n$, utilizando un tamaño de paso $h$. Este método combina información de la pendiente en cuatro puntos distintos para obtener una estimación precisa de la solución:

1. Calcular la primera pendiente (al inicio del intervalo):

$$ k_1 = h \cdot f(t_n, y_n) $$

2. Calcular la segunda pendiente (en el punto medio, usando $k_1$):

$$ k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right) $$

3. Calcular la tercera pendiente (en el punto medio, usando $k_2$):

$$ k_3 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_2}{2}\right) $$

4. Calcular la cuarta pendiente (al final del intervalo, usando $k_3$):

$$ k_4 = h \cdot f(t_n + h, y_n + k_3) $$

5. Calcular el valor siguiente de $y$ combinando todas las pendientes:

$$ \boxed{ y_{n+1} = y_n + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4) } $$

6. Avanzar en el tiempo:

$$ \boxed{ t_{n+1} = t_n + h } $$

**Nota:** Este método logra una alta precisión con un esfuerzo computacional moderado, y es ampliamente utilizado como estándar para resolver numéricamente ecuaciones diferenciales ordinarias.

```python
from numpy import ndarray
import numpy as np
from typing import Callable

def euler_method(f: Callable[[float, float], float], t0: float, y0: float, h: float, tf: float) -> tuple[ndarray, ndarray]:
    """Euler method to solve first-order ordinary differential equations (ODEs)

    Args:
        f (Callable[[float, float], float]): Function f(t, y) representing the ODE dy/dt = f(t, y)
        t0 (float): Initial value of the independent variable
        y0 (float): Initial value of the dependent variable
        h (float): Step size for the independent variable
        tf (float): Final value of the independent variable

    Returns:
        tuple[np.ndarray, np.ndarray]: Tuple (t, y) where:
            t (np.ndarray): Discrete values of the independent variable
            y (np.ndarray): Approximate values of the dependent variable at each t
    """

    # Discrete values of the independent variable
    t = np.arange(start=t0, stop=tf + h, step=h)

    # Array to store values of the dependent variable
    y = np.zeros_like(t)

    # Initial condition
    y[0] = y0

    # Iterative Euler update
    for n in range(len(t) - 1):
        y[n + 1] = y[n] + h * f(t[n], y[n])

    return t, y

def rk2_method(f: Callable[[float, float], float], t0: float, y0: float, h: float, tf: float) -> tuple[ndarray, ndarray]:
    """Euler method to solve first-order ordinary differential equations (ODEs)

    Args:
        f (Callable[[float, float], float]): Function f(t, y) representing the ODE dy/dt = f(t, y)
        t0 (float): Initial value of the independent variable
        y0 (float): Initial value of the dependent variable
        h (float): Step size for the independent variable
        tf (float): Final value of the independent variable

    Returns:
        tuple[np.ndarray, np.ndarray]: Tuple (t, y) where:
            t (np.ndarray): Discrete values of the independent variable
            y (np.ndarray): Approximate values of the dependent variable at each t
    """

    # Discrete values of the independent variable
    t = np.arange(start=t0, stop=tf + h, step=h, dtype=float)

    # Array to store values of the dependent variable
    y = np.zeros_like(t, dtype=float)

    # Initial condition
    y[0] = y0

    # Iterative RK2 update
    for n in range(len(t) - 1):
        k1 = h * f(t[n], y[n])
        t_mid = t[n] + h/2
        y_mid = y[n] + k1/2
        k2 = h * f(t_mid, y_mid)

        y[n+1] = y[n] + k2
        #t[n+1] = t[n] + h

    return t, y



def rk4_method(f: Callable[[float, float], float], t0: float, y0: float, h: float, tf: float) -> tuple[ndarray, ndarray]:
    """Runge-Kutta 4th order method (RK4) to solve first-order ODEs."""

    ######

    return t, y
```
