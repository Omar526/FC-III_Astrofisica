### **1. Dinámica del decaimiento orbital de un satélite**

Antes de estudiar el caso del decaimiento orbital, se debe considerar el caso de una órbita circular sin ningún tipo de disipación de energía mecánica.

En una órbita circular sin pérdida de energía, la gravedad actúa como fuerza centrípeta. Usando la segunda ley de Newton y la fórmula de aceleración centrípeta, se puede calcular la velocidad del satélite:

\begin{equation} v =  \sqrt{ \frac{GM}{r} } \end{equation}

donde $G$ es la constante de gravitación universal, $M$ es la masa de la Tierra, y $r$ es el radio de la trayectoria. Como resultado de la Ecuación (1), la energía mecánica conservada del sistema es:

\begin{equation} E = - \frac{GMm}{r} + \frac{1}{2} mv^2 = - \frac{GMm}{2r} \end{equation}

donde $m$ es la masa del satélite. Los valores de la constante gravitacional, la masa de la Tierra, la masa del satélite Tiangong-1 y el radio de la Tierra utilizados en este estudio se presentan en la siguiente tabla:

<div align="center">

| Símbolo | Valor                          | Unidad      | Descripción                         |
|---------|--------------------------------|-------------|-------------------------------------|
| $G$       | $6.674 \times 10^{-11}$      | N·m²/kg²    | Constante de gravitación universal  |
| $M$       | $5.972 \times 10^{24}$       | kg          | Masa de la Tierra                   |
| $m$       | $8.506 \times 10^{3}$        | kg          | Masa del satélite Tiangong-1        |
| $R_E$     | $6.378 \times 10^{6}$        | m           | Radio de la Tierra                  |

</div>


Cuando el satélite está en una órbita baja, el arrastre atmosférico reduce su energía mecánica, causando una caída en la altitud. Esta pérdida se representa como una potencia disipada proporcional al producto entre la fuerza de arrastre y la velocidad:

\begin{equation} \frac{d}{dt} \left( - \frac{GMm}{2r} \right) = \vec{F}_D \cdot \vec{v} \end{equation}

La magnitud de la fuerza de arrastre puede obtenerse a partir de la ecuación del arrastre:

\begin{equation} F_D = \frac{1}{2} \rho(r) C_D A v^2 \end{equation}

La fuerza de arrastre depende de la densidad del aire $\rho(r)$, el área del satélite $A$, el coeficiente de arrastre $C_D$ y el cuadrado de su velocidad. Dado que es una fuerza proporcional a $v^2$, no se aplica la ley de Stokes (que es válida para velocidades bajas). 

Asumiendo que la órbita sigue siendo circular mientras el satélite pierde altura lentamente, se puede combinar la ecuación del arrastre con las ecuaciones de energía y velocidad orbital para describir la decadencia de la órbita:

\begin{equation} \frac{GMm}{2} \frac{\dot{r}}{r^2} = - \frac{1}{2} \rho (r) C_D A \left( \frac{GM}{r} \right)^{3/2} \end{equation}

La simplificación de esta expresión conduce a una ecuación diferencial no lineal de primer orden para el radio de la órbita del satélite como función del tiempo:

\begin{equation} \dot{r} = -k\sqrt{r} \rho(r) \end{equation}

donde $k = \sqrt{GM} A_\text{eff}$ y $A_\text{eff} = C_D A$ es el área de sección transversal efectiva.

La misma ecuación puede escribirse en términos de la altitud del satélite $h$ sobre la superficie terrestre:

\begin{equation} \boxed{ \frac{dh}{dt} = -k \sqrt{R_E + h} \rho_0 \exp \left( - \frac{h - h_0}{H} \right) } \end{equation}

Los parámetros utilizados por los investigadores del paper son:

| $ h_0 $    | 175,000                | m              | Altitud de referencia para $ \rho_0 $             |
| $ H $      | 29,500                 | m              | Altura de escala de la atmósfera                    |
| $ A_{\text{eff}} $ | 41.8            | m²             | Área transversal efectiva del satélite ($C_D A$)  |
| $ k $      | $\sqrt{GM} \cdot A_{\text{eff}} / m$ | m/s²·kg/m² | Constante del modelo de decaimiento orbital         |
| $ t_f $    | 155 días               | s              | Tiempo final de la simulación (155 días)            |
| $ \Delta t $ | 5 días               | s              | Paso de tiempo en la simulación (5 días)            |

### **2. Solución numérica mediante Euler y RK2**

Acontinuación, implementamos los métodos Euler y RK2 en Python para estudiar el decaimiento orbital de este satélite entre $t_0 = 0$ días y $t_f=155$ días, considerando un tamaño de paso $dt = 5$ días.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
rho_0 = 6e-10  # kg/m^3
h_0 = 175_000  # m
H = 29_500     # m
G = 6.674e-11  # m^3/kg/s^2
M = 5.972e24   # kg
m = 8.506e3    # kg
R_M = 6.378e6  # m
A_eff = 41.8   # m^2 (cd * A)

# Factor k
k = np.sqrt(G * M) * A_eff / m

# Función de decaimiento
def f(t, h):
    return -k * np.sqrt(R_M + h) * rho_0 * np.exp(-(h - h_0) / H)

# Método de Euler con arange
def euler_method(f, t0, h0, tf, dt):
    t = np.arange(t0, tf + dt, dt)
    h = np.zeros_like(t, dtype=float)
    h[0] = h0

    for n in range(len(t) - 1):
        h[n+1] = h[n] + dt * f(t[n], h[n])
    
    return t, h

# Método de RK2 con arange
def rk2_method(f, t0, h0, tf, dt):
    t = np.arange(t0, tf + dt, dt)
    h = np.zeros_like(t, dtype=float)
    h[0] = h0

    for n in range(len(t) - 1):
        k1 = dt * f(t[n], h[n])
        k2 = dt * f(t[n] + dt / 2, h[n] + k1 / 2)
        h[n+1] = h[n] + k2

    return t, h

# Parámetros de simulación
tf = 155 * 24 * 60 * 60  # 155 días en segundos
dt = 5 * 24 * 60 * 60  # paso en segundos

# Simulaciones
t, h_euler = euler_method(f, 0, 300_000, tf, dt)
_, h_rk2 = rk2_method(f, 0, 300_000, tf, dt)

# Conversión de unidades
t_dias = t / (3600 * 24)
h_km_euler = h_euler / 1000
h_km_rk2 = h_rk2 / 1000

# Gráfico

# skip = 20

plt.figure(figsize=(8, 5))
plt.plot(t_dias, h_km_euler, label='Euler')
# plt.scatter(t_dias[::skip], h_km_euler[::skip], marker="o", alpha=0.3)

plt.plot(t_dias, h_km_rk2, label='RK2')
# plt.scatter(t_dias[::skip], h_km_rk2[::skip], marker="o", alpha=0.3)

plt.xlabel('Tiempo (días)')
plt.ylabel('Altura sobre la superficie (km)')
plt.title('Decaimiento Orbital')
plt.grid(ls="--")
plt.legend()
plt.tight_layout()
plt.show()
```

### **3. Estudio de convergencia de los algoritmos**

Ahora vamos a realizar un estudio de convergencia numérica para evaluar la precisión de las soluciones obtenidas con los métodos de Euler y RK2, utilizando Python. Dado que no contamos con una solución analítica, emplearemos el método de refinamiento de malla, que consiste en disminuir progresivamente el paso de tiempo $dt$ y comparar el promedio de las soluciones obtenidas para cada $dt$.

La convergencia se considera alcanzada cuando la diferencia entre los promedios de alturas de dos simulaciones consecutivas es menor que $\epsilon=0.001$ km (1 metro). Iniciaremos con un paso de 5 días y lo reduciremos sucesivamente a la mitad. Al final, compararemos los métodos de Euler y RK2 para determinar cuál requiere un paso de tiempo menor para converger.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
rho_0 = 6e-10
h_0 = 175_000
H = 29_500
G = 6.674e-11
M = 5.972e24
m = 8.506e3
R_M = 6.378e6
A_eff = 41.8
k = np.sqrt(G * M) * A_eff / m

# Ecuación de decaimiento
def f(t, h):
    return -k * np.sqrt(R_M + h) * rho_0 * np.exp(-(h - h_0) / H)

# Métodos numéricos
def euler_method(f, t0, h0, tf, dt):
    t = np.arange(t0, tf + dt, dt)
    h = np.zeros_like(t, dtype=float)
    h[0] = h0
    for n in range(len(t) - 1):
        h[n+1] = h[n] + dt * f(t[n], h[n])
    return t, h

def rk2_method(f, t0, h0, tf, dt):
    t = np.arange(t0, tf + dt, dt)
    h = np.zeros_like(t, dtype=float)
    h[0] = h0
    for n in range(len(t) - 1):
        k1 = dt * f(t[n], h[n])
        k2 = dt * f(t[n] + dt / 2, h[n] + k1 / 2)
        h[n+1] = h[n] + k2
    return t, h

# Estudio de convergencia
def convergence_study(method, name, tol=1e-3):
    print(f"\nEstudio de convergencia para {name}")
    tf = 155 * 24 * 3600  # 155 días
    h0 = 300_000
    t0 = 0
    dt = 5 * 24 * 60 * 60  # 5 días en segundos
    h_prom_anterior = None
    converged = False

    dts = []
    h_proms = []
    epsilons = []

    while not converged:
        t, h = method(f, t0, h0, tf, dt)
        h_prom = np.mean(h) / 1000  # en km

        dts.append(dt)
        h_proms.append(h_prom)

        if h_prom_anterior is not None:
            eps = abs(h_prom - h_prom_anterior)
            epsilons.append(eps)
            print(f"dt = {dt:6.1f},  h_prom = {h_prom:.4f} km,  ε = {eps:.5f}")
            if eps < tol:
                print(f"✅ Convergencia alcanzada para {name} con dt = {dt:.1f} s")
                break
        else:
            epsilons.append(np.nan)
            print(f"dt = {dt:6.1f},  h_prom = {h_prom:.4f} km,  ε = ---")

        h_prom_anterior = h_prom
        dt /= 2

    return dts, h_proms, epsilons

# Ejecutar y graficar convergencia
dts_euler, h_euler, eps_euler = convergence_study(euler_method, "Euler")
dts_rk2, h_rk2, eps_rk2 = convergence_study(rk2_method, "RK2")

# Graficar
plt.figure(figsize=(8, 5))

plt.plot(dts_euler, eps_euler, '-o', label='Euler')
plt.plot(dts_rk2, eps_rk2, '-o', label='RK2')
plt.axhline(0.001, color='gray', linestyle='--', label='Tolerancia 0.01 km')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('dt (s)')
plt.ylabel('ε (km)')
plt.title('Convergencia ε vs dt')
plt.grid(True, which='both', ls='--')
plt.legend()


plt.tight_layout()
plt.show()
```

Luego de realizar el estudio de convergencia, se determinaron los tamaños de paso óptimos para cada método numérico: $dt=52.7$ segundos para Euler y $dt=210.9$ segundos para RK2. Utilizando estos valores, graficamos la solución de la ecuación diferencial, obteniendo así una aproximación precisa de la evolución de la altura del satélite a lo largo del tiempo:

```python
# Parámetros de simulación
tf = 155 * 24 * 60 * 60  # 155 días en segundos

# Simulaciones
t_euler, h_euler = euler_method(f, 0, 300_000, tf, dt=52.7)
t_rk2, h_rk2 = rk2_method(f, 0, 300_000, tf, dt=210.9)

# Conversión de unidades
t_dias_euler = t_euler / (3600 * 24)
h_km_euler = h_euler / 1000

t_dias_rk2= t_rk2 / (3600 * 24)
h_km_rk2 = h_rk2 / 1000

# Gráfico

# skip = 20

plt.figure(figsize=(8, 5))
plt.plot(t_dias_euler, h_km_euler, label='Euler')
# plt.scatter(t_dias[::skip], h_km_euler[::skip], marker="o", alpha=0.3)

plt.plot(t_dias_rk2, h_km_rk2, label='RK2')
# plt.scatter(t_dias[::skip], h_km_rk2[::skip], marker="o", alpha=0.3)

plt.xlabel('Tiempo (días)')
plt.ylabel('Altura sobre la superficie (km)')
plt.title('Decaimiento Orbital')
plt.grid(ls="--")
plt.legend()
plt.tight_layout()
plt.show()
```

### **5. Pérdida de Energía y Trabajo asociado a la Fuerza de Arrastre**

Vamos a calcular la pérdida de energía del satélite causada por la fuerza de arrastre, que al ser una fuerza no conservativa realiza trabajo sobre el sistema. Para ello, implementaremos en Python una función que utilice el método trapezoidal discreto para integrar el trabajo realizado por la fuerza de arrastre a lo largo del tiempo. Posteriormente, compararemos este valor con la variación de la energía mecánica total del satélite, calculada mediante la fórmula de energía orbital, para verificar si la disminución de energía coincide aproximadamente con el trabajo realizado por el arrastre.

```python
def trapezoidal_integral(x, y):
    """
    Aproxima la integral definida ∫f(x) dx usando el método del trapecio
    para datos discretos.

    Parámetros:
        x: array-like, puntos en el eje x (deben estar ordenados)
        y: array-like, valores f(x) en los puntos x

    Retorna:
        Aproximación de la integral (float)
    """
    x = np.asarray(x)
    y = np.asarray(y)
    dx = np.diff(x)
    avg_y = (y[:-1] + y[1:]) / 2
    return np.sum(avg_y * dx)


v = np.sqrt(G * M / (R_M + h_rk2))
F_D = 0.5 * rho_0 * np.exp(-(h_rk2 - h_0) / H) * A_eff * v**2

integrand = - F_D * v

W = trapezoidal_integral(x=t_rk2, y=integrand)

print(f"Trabajo realizado por la fuerza de arrastre: {W:.2e} J = {W*1e-9:.2f} GJ")

# Radios correspondientes
r1 = R_M + h_rk2[0]
r2 = R_M + h_rk2[-1]

# Energías orbitales en h1 y h2
E1 = - G * M * m / (2 * r1)
E2 = - G * M * m / (2 * r2)
delta_E = E2 - E1  # Cambio de energía (negativo)

print(f"Cambio de energía orbital (E2 - E1): {delta_E * 1e-9:.2f} GJ")
```

El trabajo realizado por la fuerza de arrastre es de aproximadamente $−6.20$ GJ. Este valor coincide con el cambio en la energía orbital del satélite, calculado como la diferencia entre la energía final y la inicial. Por lo tanto, se confirma que la pérdida de energía del satélite se debe efectivamente al trabajo realizado por la fuerza de arrastre.

### **6. Estimación Energética para Corrección Orbital**

Vamos a resolver este problema calculando el tiempo máximo que puede transcurrir antes de realizar una maniobra de corrección orbital, considerando que el sistema de propulsión dispone de un máximo de 600 MJ de energía. Para ello, sumaremos el trabajo acumulado realizado por la fuerza de arrastre a lo largo del tiempo, paso a paso, hasta que dicho trabajo alcance ese límite energético. El tiempo en que se alcanza este valor marcará el máximo tiempo de espera, así como el máximo descenso permitido en la altura del satélite antes de ejecutar la maniobra. Utilizaremos un ciclo iterativo en Python para realizar esta suma y detener el cálculo en el momento oportuno.

```python

W_acum = 0
dt = t_rk2[1] - t_rk2[0]

Wf = 600e6 # J
i = 0
while W_acum <= Wf:
    W_acum += F_D[i] * v[i] * dt
    i += 1

    if i >= len(t_rk2):
        break


print(f"{round(t_rk2[i]/(60 * 60 * 24))} días")

print(f"{round(h_km_rk2[0] - h_km_rk2[i])} km")
```

Después de 64 días, el satélite habrá consumido la totalidad de los 600 MJ disponibles para la maniobra de corrección orbital, lo que corresponde a una pérdida de altitud de aproximadamente 16 km.
